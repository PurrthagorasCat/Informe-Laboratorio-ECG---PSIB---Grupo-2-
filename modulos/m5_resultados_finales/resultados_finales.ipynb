## Ítem 8 — Material para la discusión

- **FC**: sube de 84 a 131 lpm de media después del ejercicio y baja ~26 lpm durante los 3 min de registro. Es lo
  esperado: retirada vagal y activación simpática durante el esfuerzo, y recuperación con reactivación vagal rápida
  seguida de una caída simpática más lenta.
- **Variabilidad**: RMSSD de 48 a 8,6 ms; LF y HF caen entre 50 y 100 veces. Coincide con la bibliografía (la VFC
  cae con la FC y con el ejercicio). Parte de la caída es "mecánica": con RR más cortos, la misma modulación relativa
  da variaciones absolutas menores (por eso se miran también las unidades normalizadas).
- **Minuto a minuto**: en el primer minuto HF casi desaparece y el pico respiratorio se va a 0,36 Hz (respiración
  agitada); en los minutos 2 y 3 HF vuelve. Coincide con la reactivación vagal gradual.
- **Reposo**: LF/HF = 4,7, alto para reposo. El registro (55 s) es más corto que los 2 min que pide la Task Force para
  LF y está dominado por pocas oscilaciones lentas grandes; poco confiable.
- **Limitaciones**: un solo voluntario (sin ítems 6b/7b), registros cortos, sin registro de la respiración, fs con
  huecos entre paquetes (error sistemático del 4 % en el eje de tiempo), tiempo desconocido entre el fin del ejercicio
  y el inicio del registro.
